```python
import pandas
```

```python
# read expression
tpm_file = '/Users/adrian/research/bmcbf/061_drangajokull/profiles/DESeq2_TPM_values.transcript_level.tsv'
```

```python
df = pandas.read_csv(tpm_file, sep='\t', index_col=0)
df
```

```python
# read annotation
annotation_file = '/Users/adrian/software/kallisto/human_index_standard/t2g.txt'
annotation = pandas.read_csv(annotation_file, sep='\t', header=None)
sub_anno = annotation[annotation[2] == 'ATG7']
print(sub_anno.shape)
sub_anno
transcript_isoforms = sub_anno[0]
```

```python
# define plotting data
plot_data = df.loc[transcript_isoforms, :]
plot_data['names'] = sub_anno[3].values
plot_data.set_index('names', inplace=True)
plot_data
```

```python
# plot
matplotlib.pyplot.imshow(plot_data, aspect='auto', cmap='viridis')

matplotlib.pyplot.xticks(range(len(plot_data.columns)), plot_data.columns, rotation=90)
matplotlib.pyplot.yticks(range(len(plot_data.index)), plot_data.index)

cbar = matplotlib.pyplot.colorbar()
cbar.set_label("TPM", fontsize=12)

matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```

```python
# 1) Build a mask based on triplicate medians
group_size = 3
num_columns = len(plot_data.columns)

group_medians = []

for start in range(0, num_columns, group_size):
    group = plot_data.iloc[:, start:start + group_size]   # 3 columns
    # median across the 3 replicates for each row
    group_medians.append(group.median(axis=1))

# Combine all medians into one DataFrame (rows x conditions)
median_df = group_medians[0].to_frame(name="cond_1")
for index, series in enumerate(group_medians[1:], start=2):
    median_df[f"cond_{index}"] = series

# Row is kept if ANY condition median > 1
row_mask = (median_df > 1).any(axis=1)

# 2) Filter original data using this mask (keep all columns)
mini_df = plot_data[row_mask]

matplotlib.pyplot.imshow(mini_df, aspect='auto', cmap='cividis', vmin=1)

# Set xticks
xticks = range(len(mini_df.columns))
matplotlib.pyplot.xticks(xticks, mini_df.columns, rotation=90)

# Color labels depending on substring
for tick, col in zip(matplotlib.pyplot.gca().get_xticklabels(), mini_df.columns):
    if "siA" in col:
        tick.set_color("red")   # highlight siA
    else:
        tick.set_color("black") # or remove this line to leave default

# y-ticks unchanged
matplotlib.pyplot.yticks(range(len(mini_df.index)), mini_df.index)

# Colorbar
cbar = matplotlib.pyplot.colorbar()
cbar.set_label("TPM", fontsize=12)

matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```

```python
from scipy import stats  # add this somewhere at the top of your script

subset_rows = mini_df.index
group_size = 3
num_columns = len(mini_df.columns)

for gene in subset_rows:
    row = mini_df.loc[gene]

    groups = []
    group_labels = []
    print(f"\nGene: {gene}")

    for start in range(0, num_columns, group_size):
        trip_vals = row.iloc[start:start + group_size].values
        groups.append(trip_vals)

        col_names = list(mini_df.columns[start:start + group_size])

        # base label from first name minus last character (A1 → A, C1 → C)
        base_label = col_names[0][:-1]
        group_labels.append(base_label)

        composite_label = f"{base_label} ({', '.join(col_names)})"
        print(f"  {composite_label}: {trip_vals}")

    # one figure per gene
    matplotlib.pyplot.figure()

    # boxplot (use tick_labels per Matplotlib's new API)
    boxdict = matplotlib.pyplot.boxplot(
        groups,
        tick_labels=group_labels,
        patch_artist=True
    )

    # color boxes as before
    condition_colors = [
        "tab:red", "tab:red",
        "tab:orange", "tab:orange",
        "white", "white",
        "skyblue", "skyblue"
    ]

    for i, box in enumerate(boxdict["boxes"]):
        fill_color = condition_colors[i] if i < len(condition_colors) else "white"
        box.set_facecolor(fill_color)
        box.set_edgecolor("black")

    for median in boxdict["medians"]:
        median.set_color("black")

    for whisker in boxdict["whiskers"]:
        whisker.set_color("black")

    for cap in boxdict["caps"]:
        cap.set_color("black")

    for flier in boxdict["fliers"]:
        flier.set(markeredgecolor="black")

    # scatter points over boxes
    for index, trip_vals in enumerate(groups, start=1):
        x_positions = [index - 0.2, index, index + 0.2][:len(trip_vals)]
        matplotlib.pyplot.scatter(
            x_positions,
            trip_vals,
            color="black",
            s=20,
            zorder=3
        )

    # ---------------------------------
    # Pairwise t-tests and stars
    # ---------------------------------
    for pair_start in range(0, len(groups), 2):
        if pair_start + 1 >= len(groups):
            break

        g1 = groups[pair_start]
        g2 = groups[pair_start + 1]

        t_stat, p_value = stats.ttest_ind(g1, g2, equal_var=True)
        print(
            f"  t-test {group_labels[pair_start]} vs {group_labels[pair_start+1]}: "
            f"t = {t_stat:.3f}, p = {p_value:.3e}"
        )

        if p_value < 0.05:
            # x positions of the two boxes in this pair: 1-based (boxplot convention)
            x1 = pair_start + 1
            x2 = pair_start + 2
            x_star = x1   # at the si

            matplotlib.pyplot.text(
                x_star,
                -2,
                "*",
                ha="center",
                va="center",
                fontsize=28,
                color="black",
                clip_on=False
            )

    # y-range and grid
    matplotlib.pyplot.ylim(-4, 27)
    matplotlib.pyplot.grid(alpha=0.5, ls=":")

    matplotlib.pyplot.ylabel("TPM")
    matplotlib.pyplot.title(str(gene))

    matplotlib.pyplot.tight_layout()
    matplotlib.pyplot.show()
```
