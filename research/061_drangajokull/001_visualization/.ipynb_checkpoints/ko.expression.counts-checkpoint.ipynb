```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
import pandas
```

```python
# read expression
tpm_file = '/Users/adrian/research/bmcbf/061_drangajokull/profiles/DESeq2_TPM_values.transcript_level.tsv'
```

```python
df = pandas.read_csv(tpm_file, sep='\t', index_col=0)
df
```

```python
# read annotation
annotation_file = '/Users/adrian/software/kallisto/human_index_standard/t2g.txt'
annotation = pandas.read_csv(annotation_file, sep='\t', header=None)
sub_anno = annotation[annotation[2] == 'ATG7']
print(sub_anno.shape)
sub_anno
transcript_isoforms = sub_anno[0]
```

```python
# define plotting data
plot_data = df.loc[transcript_isoforms, :]
plot_data['names'] = sub_anno[3].values
plot_data.set_index('names', inplace=True)
plot_data
```

```python
# plot
matplotlib.pyplot.imshow(plot_data, aspect='auto', cmap='viridis')

matplotlib.pyplot.xticks(range(len(plot_data.columns)), plot_data.columns, rotation=90)
matplotlib.pyplot.yticks(range(len(plot_data.index)), plot_data.index)

cbar = matplotlib.pyplot.colorbar()
cbar.set_label("TPM", fontsize=12)

matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```

```python
mini_df = plot_data[plot_data.gt(1).any(axis=1)]

matplotlib.pyplot.imshow(mini_df, aspect='auto', cmap='viridis')

# Set xticks
xticks = range(len(mini_df.columns))
matplotlib.pyplot.xticks(xticks, mini_df.columns, rotation=90)

# Color labels depending on substring
for tick, col in zip(matplotlib.pyplot.gca().get_xticklabels(), mini_df.columns):
    if "siA" in col:
        tick.set_color("red")   # highlight siA
    else:
        tick.set_color("black") # or remove this line to leave default

# y-ticks unchanged
matplotlib.pyplot.yticks(range(len(mini_df.index)), mini_df.index)

# Colorbar
cbar = matplotlib.pyplot.colorbar()
cbar.set_label("TPM", fontsize=12)

matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```
