```python
import pandas, supervenn
```

```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
input_dir = '/Users/adrian/research/bmcbf/025_isafjordur/results/002_DEGs/'

df = pandas.read_csv(input_dir+'effect_ko_vs_wt.skmel.for.tsv', sep='\t', index_col=0)
a = set(df.index)
print(len(a))

df = pandas.read_csv(input_dir+'effect_ko_vs_wt.u2os.for.tsv', sep='\t', index_col=0)
b = set(df.index)
print(len(b))

df = pandas.read_csv(input_dir+'effect_ko_vs_wt.u2os.group.for.tsv', sep='\t', index_col=0)
c = set(df.index)
print(len(c))

df = pandas.read_csv(input_dir+'effect_ko_vs_wt.u2os.two.for.tsv', sep='\t', index_col=0)
d = set(df.index)
print(len(d))
```

```python
sets = [a, b, c, d]
labels = ['skmel', 'u2os n3', 'u2os groups', 'u2os n2']
mso = supervenn.supervenn(sets, labels, chunks_ordering='size',
                          side_plots=False,
                          reverse_sets_order=False, fontsize=8, color_cycle=['tab:red', 'skyblue', 'tab:green', 'gold'])

matplotlib.pyplot.show()
```
