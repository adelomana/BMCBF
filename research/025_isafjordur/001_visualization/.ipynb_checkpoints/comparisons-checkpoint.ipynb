```python
import pandas, matplotlib_venn
```

```python
import matplotlib
```

```python
skmel_degs_file = '/Users/adrian/research/bmcbf/025_isafjordur/results/002_DEGs/effect_ko_vs_wt.skmel.for.tsv'
u2os_degs_file = '/Users/adrian/research/bmcbf/025_isafjordur/results/002_DEGs/effect_ko_vs_wt.u2os.two.for.tsv'
output_dir = '/Users/adrian/research/bmcbf/025_isafjordur/results/002_DEGs/'
```

```python
dfskmel = pandas.read_csv(skmel_degs_file, sep='\t')
skmel_degs = set(list(dfskmel.index))
print(len(skmel_degs))

dfu2os = pandas.read_csv(u2os_degs_file, sep='\t')
u2os_degs = set(list(dfu2os.index))
print(len(u2os_degs))
```

```python
results = matplotlib_venn.venn2(subsets = [skmel_degs, u2os_degs], set_labels=['skmel', 'u2os'], set_colors=['tab:orange', 'skyblue'], alpha=0.7)
matplotlib_venn.venn2_circles(subsets=[skmel_degs, u2os_degs])  

#matplotlib.pyplot.show()
matplotlib.pyplot.savefig('venn.svg')
```

```python
import supervenn
```

```python
sets = [skmel_degs, u2os_degs]
labels = ['skmel', 'u2os']
mso = supervenn.supervenn(sets, labels,
                          side_plots=False, widths_minmax_ratio=0.025,
                          reverse_sets_order=False, fontsize=8, color_cycle=['tab:orange', 'skyblue'])

matplotlib.pyplot.show()
```

```python
skmelu = list(mso.get_chunk([0, 0]))
intersect = list(mso.get_chunk([0, 1]))
u2osu = list(mso.get_chunk([1,1]))
print(len(skmelu))
print(len(intersect))
print(len(u2osu))
```

```python
df = dfskmel.loc[skmelu]
print(df.shape)
df.to_csv(output_dir+'set_skmelu.csv', sep='\t')

df = dfskmel.loc[intersect]
print(df.shape)
df.to_csv(output_dir+'set_intersect.csv', sep='\t')

df = dfu2os.loc[u2osu]
print(df.shape)
df.to_csv(output_dir+'set_u2osu.csv', sep='\t')
```
