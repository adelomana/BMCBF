```python
import pandas, os
```

```python
input_folder = '/Users/adrian/research/019.fossvogur/001_profiles/'

# a cases
a_cases = ['SkMel28-MITFKO_ev_skmel28_rep1', 'SkMel28-MITFKO_ev_skmel28_rep2', 'SkMel28-MITFKO_ev_skmel28_rep3', 'SkMel28-MITFKO_ev_skmel28_rep4', 
          'SkMel28-MITFKO_mitf_x6_rep1', 'SkMel28-MITFKO_mitf_x6_rep3', 'SkMel28-MITFKO_mitf_x6_rep4']
```

```python
samples = os.listdir(input_folder)
samples.remove('messages')
samples.sort()
print(len(samples), samples) # should be 30 in this case
```

```python
expression = pandas.DataFrame()

for sample in samples[:1]:
    print('working with sample {}'.format(sample))

    label = sample.replace('_processed', '')
    if label in a_cases:
        target_file = input_folder + sample + '/kallisto_output_a' + '/abundance.tsv'
    else:
        target_file = input_folder + sample + '/kallisto_output_c' + '/abundance.tsv'
    print('\t {}'.format(target_file))

    df = pandas.read_csv(target_file, sep='\t', index_col=0)

    expression.join(df['tpm'], how='left')

df
```
