```python
# get cannonical HH genes based on R-HSA Hedgehog 'On' State term
with open('Reactome_Pathways_2024.txt', newline='') as f:
    reader = csv.reader(f, delimiter='\t')   # adjust delimiter
    for row in reader:
        if row[0] == "Hedgehog 'On' State":
            canonical = list(set(row[2:-1]))
            print(row, len(row))
            print(canonical, len(canonical))   
```

```python
# get the non cannonical hh based on Table 1 from 

def appender(noncanonical):
    elements = set(row[2:-1])
    print('detected category with', len(elements))
    noncanonical = noncanonical | elements
    print('current', len(noncanonical))
    print()
    return noncanonical

noncanonical = set()

categories = [
    "RAF MAP Kinase Cascade", 'MAPK1 MAPK3 Signaling', 'MAPK1 (ERK2) Activation', 'MAPK3 (ERK1) Activation'
]

with open('Reactome_Pathways_2024.txt', newline='') as f:
    reader = csv.reader(f, delimiter='\t')   # adjust delimiter
    for row in reader:
        
        # RAS-RAF-MEK-ERK: RAF MAP Kinase Cascade, 

        for category in categories:
            if category == row[0]:
                noncanonical = appender(noncanonical)
            
        # MAPK1 MAPK3 Signaling
            
noncanonical = list(noncanonical)  
print(len(noncanonical), noncanonical[:10])
```

```python
# convert the two lists into ensembl
t2g_file = '/Users/adrian/software/kallisto/human_index_standard/t2g.txt'
rosetta = pandas.read_csv(t2g_file, sep='\t', header=None)

ensembl_cannonical = []; ensembl_noncanonical = []

def converter(target):
    ensemblID = None
    conversion = list(set(rosetta[rosetta[2] == target][1].values))
    if len(conversion) == 0:
        pass
    elif len(conversion) == 1:
        ensemblID = conversion[0]
    else:
        print('ERROR')
        print(conversion)
    return ensemblID

for target in canonical:
    ensemblID = converter(target)
    if ensemblID is not None:
        ensembl_cannonical.append(ensemblID)
   
for target in noncanonical:
    ensemblID = converter(target)
    if ensemblID is not None:
        ensembl_noncanonical.append(ensemblID)

ensembl_cannonical = list(set(ensembl_cannonical))
ensembl_noncanonical = list(set(ensembl_noncanonical))
print(len(ensembl_cannonical), len(ensembl_noncanonical))

targets = list(set(ensembl_cannonical + ensembl_noncanonical))
print(len(targets))
```

```python
# access the expression data
expression_file = 'DESeq2_TPM_values.transcripts.tsv'
df = pandas.read_csv(expression_file, sep='\t', index_col=0)
df
df.shape
```

```python
# subset based on expression
sub = df.loc[targets]
sub.shape
```

```python
# exclude a few samples because they were DOX treated: SRR24197676, SRR24197677, SRR24197678
sub = sub.drop(columns=['SRR24197676', 'SRR24197677', 'SRR24197678'])
sub.shape
```

```python
# they seem to be TPMs
sub.head()
```

```python
# get the top variant genes based on log2 transformed MAD
x = numpy.log2(sub + 1.0)
mad = x.sub(x.median(axis=1), axis=0).abs().median(axis=1)
top_genes = mad.sort_values(ascending=False).head(300).index # get the the top 100 most different genes
x_top = x.loc[top_genes]
```

```python
# row-wise z-score
row_mean = x_top.mean(axis=1)
row_std = x_top.std(axis=1, ddof=1).replace(0.0, numpy.nan)
z = x_top.sub(row_mean, axis=0).div(row_std, axis=0).fillna(0.0)
```

```python
# bring in gene names
new_names = []
for element in z.index:
    new_names.append(list(set(rosetta[rosetta[1] == element][2]))[0])
print(len(z.index), len(new_names))

z.index = new_names
```

```python
# add metadata
metadata_file = 'SraRunTable.csv'
metadata = pandas.read_csv(metadata_file, sep=',')

# Remove doxycycline-treated A-673 samples
metadata = metadata[~metadata['Run'].isin(['SRR24197676', 'SRR24197677', 'SRR24197678'])]

# Clean cell line names
run_to_cellline = dict(zip(metadata['Run'], metadata['cell_line']))
run_to_cellline = {k: v.split(' (')[0] for k, v in run_to_cellline.items()}

# Ewing and osteo lines after removal
ewing_lines = metadata[metadata['cell_type'] == 'ewing sarcoma']['cell_line'].str.split(' (', regex=False).str[0].unique()
osteo_lines = metadata[metadata['cell_type'] == 'osteosarcoma']['cell_line'].str.split(' (', regex=False).str[0].unique()

ewing_colors = [
    "#FF7F0E",  # A-673       - classic orange
    "#E8564A",  # EW-24       - coral red
    "#C2557A",  # EW-3        - raspberry
    "#9B59B6",  # MHH-ES-1    - violet
    "#E91E8C",  # RD-ES       - hot pink
    "#FF69B4",  # SK-ES-1     - soft pink
    "#D4A017",  # TC-71       - golden amber
]

osteo_colors = [
    "#1B7A4A",  # G-292       - deep forest green
    "#2ECC71",  # KHOS/NP     - emerald
    "#17A589",  # MG-63       - deep teal
    "#1ABC9C",  # MNNG/HOS    - teal
    "#A8D8A8",  # SaOS-2      - soft sage (lighter, distinct)
    "#B8D4E8",  # U2OS        - steel blue-grey (clearly off-green, distinct)
]

cell_line_palette = {}
for i, cl in enumerate(sorted(ewing_lines)):
    cell_line_palette[cl] = ewing_colors[i % len(ewing_colors)]
for i, cl in enumerate(sorted(osteo_lines)):
    cell_line_palette[cl] = osteo_colors[i % len(osteo_colors)]

# Build col_colors aligned to z.columns
col_colors = pandas.Series(z.columns, index=z.columns).map(run_to_cellline).map(cell_line_palette)

# Row colors
row_group = pandas.Series("Other", index=z.index, dtype="object")
row_group.loc[row_group.index.isin(canonical)] = "HH_canonical"
row_group.loc[row_group.index.isin(noncanonical)] = "HH_noncanonical"
row_palette = {
    "HH_canonical": "skyblue",
    "HH_noncanonical": "gold",
}
row_colors = row_group.map(row_palette)
```

```python
# euclidean ward seems to work fine

g = seaborn.clustermap(
    z,
    metric='euclidean',
    method='ward',
    cmap='bwr',
    standard_scale=None,
    figsize=(9, 14),
    col_colors=col_colors,
    row_colors=row_colors,
    dendrogram_ratio=(0.1, 0.1),
    cbar_pos=(-0.1, 0.8, 0.05, 0.18),
    xticklabels=True,
    cbar_kws={'label': 'Z-score'},
    vmin=-2.5,
    vmax=2.5,
    yticklabels=True
)

handles = []

# Cell line legend
for label, color in cell_line_palette.items():
    handles.append(
        matplotlib.patches.Patch(
            facecolor=color,
            edgecolor='none',
            label=label
        )
    )

# Gene group legend
for label, color in row_palette.items():
    handles.append(
        matplotlib.patches.Patch(
            facecolor=color,
            edgecolor='none',
            label=f"Genes: {label}"
        )
    )

g.ax_heatmap.legend(
    handles=handles,
    loc="upper left",
    bbox_to_anchor=(1.15, 1.0),
    frameon=False,
    title="Annotations"
)

matplotlib.pyplot.show()
```
