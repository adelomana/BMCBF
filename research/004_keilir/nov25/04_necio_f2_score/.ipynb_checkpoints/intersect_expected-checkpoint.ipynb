```python
import pandas
```

```python
input_file = "db/ChEA_2022.txt"

with open(input_file) as f:
    header = f.readline().rstrip("\n").split("\t")
   

    for i, line in enumerate(f):
        if i >= 999999:
            break
        fields = line.rstrip("\n").split("\t")
        if 'MITF' in fields[0]:
            targets = fields[1:]
            
print(len(targets))
targets = [x for x in targets[1:] if x.strip()]
print(len(targets))
known_targets = list(set(targets))
print(len(targets))
```

```python
# Minimal set-based metrics: F1, F2, Jaccard
# Inputs are Python sets of gene symbols

def metrics(known, predicted):
    known = set(known)
    predicted = set(predicted)
    print(len(known), len(predicted))

    tp = len(known & predicted)
    fp = len(predicted - known)
    fn = len(known - predicted)

    precision = tp / (tp + fp) if (tp + fp) else 0.0
    recall    = tp / (tp + fn) if (tp + fn) else 0.0

    f1 = (2 * precision * recall / (precision + recall)) if (precision + recall) else 0.0
    f2 = (5 * precision * recall / (4 * precision + recall)) if (4 * precision + recall) else 0.0

    jaccard = tp / len(known | predicted) if (known | predicted) else 0.0

    print(tp, known & predicted)
    print({
        "precision": precision,
        "recall": recall,
        "F1": f1,
        "F2": f2,
        "jaccard": jaccard,
    })

    return None
```

```python
### get the genes from called peaks
input_file = '1kb/results_orange/genes_with_promoter_DBP.orange.txt'
df = pandas.read_csv(input_file, sep='\t')
identified = set(df['Gene'].to_list())
print(len(identified))
metrics(known_targets, identified)
print()

input_file = '1kb/results_purple/genes_with_promoter_DBP.purple.txt'
df = pandas.read_csv(input_file, sep='\t')
identified = set(df['Gene'].to_list())
print(len(identified))
metrics(known_targets, identified)
print()

input_file = '2kb/results_orange/genes_with_promoter_DBP.orange.txt'
df = pandas.read_csv(input_file, sep='\t')
identified = set(df['Gene'].to_list())
print(len(identified))
metrics(known_targets, identified)
print()

input_file = '2kb/results_purple/genes_with_promoter_DBP.purple.txt'
df = pandas.read_csv(input_file, sep='\t')
identified = set(df['Gene'].to_list())
print(len(identified))
metrics(known_targets, identified)
```
