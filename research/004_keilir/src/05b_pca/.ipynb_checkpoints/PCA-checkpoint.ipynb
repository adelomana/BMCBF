```python
import pandas
import sklearn.decomposition
import sklearn.preprocessing
import matplotlib.pyplot
```

```python
input_file = '/Users/adrian/research/bmcbf/004_keilir/results/05_elja_count_retriever/counts_featureCounts.txt'
```

```python
df = pandas.read_csv(input_file, sep='\t', skiprows=1)
df.head()
```

```python
df.columns = [f"{m}_{r}" if pandas.notna(m) else c for c, (m, r) in zip(df.columns, df.columns.to_series().str.extract(r"(MITF_[AM]).*?_(\d+)").to_numpy())]
df.head()
```

```python
df_mitf = df.loc[:, df.columns.str.startswith("MITF_")]
df_mitf
```

```python
substantial_expression = df_mitf[df_mitf.max(axis=1) >= 2]
high_expression = df_mitf[df_mitf.max(axis=1) >= 100]

print(substantial_expression.shape)
print(high_expression.shape)
```

```python
X = df_mitf.transpose()  # rows = samples, columns = features
X_scaled = sklearn.preprocessing.StandardScaler().fit_transform(X)

pca = sklearn.decomposition.PCA(n_components=2)
pcs = pca.fit_transform(X_scaled)

pca_df = pandas.DataFrame(pcs, index=X.index, columns=["PC1", "PC2"])
pca_df
```

```python
matplotlib.pyplot.figure(figsize=(6, 6))
matplotlib.pyplot.scatter(pca_df["PC1"], pca_df["PC2"], s=120)

for name in pca_df.index:
    matplotlib.pyplot.text(
        pca_df.loc[name, "PC1"],
        pca_df.loc[name, "PC2"],
        name,
        fontsize=12
    )

matplotlib.pyplot.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
matplotlib.pyplot.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```
