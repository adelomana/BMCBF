```python
import pandas
```

```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
parent_dir = '/Users/adrian/research/bmcbf/004_keilir/results/001_bam/'
for i in range(3):
    sample_name = 'MITF_A_Untreated_FLAG_{}'.format(i+1)
    inputfile = parent_dir + sample_name + '/fragment_lengths.txt'
    df = pandas.read_csv(inputfile, sep='\t', header=None)
    x = df[0]; y = df[1]
    matplotlib.pyplot.plot(x, y, color='gold', label='flagged')

    sample_name = 'MITF_A_Untreated_IgG_{}'.format(i+1)
    inputfile = parent_dir + sample_name + '/fragment_lengths.txt'
    df = pandas.read_csv(inputfile, sep='\t', header=None)
    x = df[0]; y = df[1]
    matplotlib.pyplot.plot(x, y, color='skyblue', label='IgG')

matplotlib.pyplot.xlim(20, 100)
matplotlib.pyplot.grid(ls=':')
handles, labels = matplotlib.pyplot.gca().get_legend_handles_labels()
matplotlib.pyplot.legend(handles[:2], labels[:2])
matplotlib.pyplot.title('MITF A')
matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```
