```python
import pandas
```

```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
parent_dir = '/Users/adrian/research/bmcbf/004_keilir/results/001b_bedgraph/results_000_0925/'
for i in range(3):
    sample_name = 'MITF_A_Untreated_FLAG_{}'.format(i+1)
    inputfile = parent_dir + sample_name + '/fragment_length_info.txt'
    df = pandas.read_csv(inputfile, sep='\t', header=None)
    x = df[0]; y = df[1]

    if i == 0:
        the_color = 'gold'; the_label = 'flagged rep. 1'
    if i == 1:
        the_color = 'tab:orange'; the_label = 'flagged rep. 2'
    if i == 2:
        the_color = 'tab:red'; the_label = 'flagged rep. 3'
    matplotlib.pyplot.plot(x, y, color=the_color, label=the_label, lw=2)

    sample_name = 'MITF_A_Untreated_IgG_{}'.format(i+1)
    inputfile = parent_dir + sample_name + '/fragment_length_info.txt'
    df = pandas.read_csv(inputfile, sep='\t', header=None)
    x = df[0]; y = df[1]
    matplotlib.pyplot.plot(x, y, color='skyblue', label='IgG', lw=2)

matplotlib.pyplot.xlim(20, 300)
matplotlib.pyplot.grid(ls=':')
handles, labels = matplotlib.pyplot.gca().get_legend_handles_labels() 

matplotlib.pyplot.legend([handles[i] for i in [0, 2, 4, 5]], [labels[i] for i in [0, 2, 4, 5]])
matplotlib.pyplot.title('MITF A')
matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```
