```python
import pandas, os, datetime
```

```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
parentdir = '/Users/adrian/research/bmcbf/004_keilir/results/001b_bedgraph/results_001'
folders = os.listdir(parentdir)
folders.sort()
```

```python
all_locations = set()
for folder in folders:
    print(datetime.datetime.now().strftime("%H:%M:%S"))
    print(folder)
    path = parentdir + f"/{folder}/human.fragments.500.bed" 
    print(path)
    df = pandas.read_csv(path, sep='\t', header=None)
    print(df.shape)
    locations = set(df[0].astype(str) + "_" + df[1].astype(str))
    all_locations.update(locations)
    print(len(all_locations))
    print()
```

```python
info = pandas.DataFrame(columns=['chr', 'loc'], index=list(all_locations))
for i in range(len(folders)):
    folder = folders[i]
    print(folder)
    path = parentdir + f"/{folder}/human.fragments.500.bed" 
    df = pandas.read_csv(path, sep='\t', header=None)
    df.index = df[0].astype(str) + "_" + df[1].astype(str)
    info[folder] = df[2]     
info.fillna(0, inplace=True)

info = info.drop(columns=["chr", "loc"])
print(info.shape)
info
```

```python
threshold = 1000
substantial = info[info.ge(threshold).any(axis=1)]
substantial.shape
substantial
```

```python
# for idx, row in substantial.iterrows():
#     min_val = row.min()
#     print(f"Row {idx} -> min value = {min_val}")
```

```python
corr = substantial.corr(method="spearman")
```

```python
matplotlib.pyplot.imshow(corr, cmap="viridis", vmin=0, vmax=1)
matplotlib.pyplot.colorbar(fraction=0.046, pad=0.04, label='SCC')

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        matplotlib.pyplot.text(j, i, f"{corr.iloc[i, j]:.2f}",
                 ha="center", va="center", color="black", fontsize=6)

names = ['Af1', 'Af2', 'Af3', 'A.1', 'A.2', 'A.3', 'Mf1', 'Mf2', 'Mf3', 'M.1', 'M.2', 'M.3']
matplotlib.pyplot.xticks(range(len(corr.columns)), names, rotation=90)
matplotlib.pyplot.yticks(range(len(corr.columns)), names)

matplotlib.pyplot.tight_layout()
matplotlib.pyplot.show()
```
