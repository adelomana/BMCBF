```python
import pandas, numpy
```

```python
import numpy, matplotlib_venn
```

```python
import matplotlib, matplotlib.pyplot
matplotlib.rcParams.update({'font.size':20, 
                            'font.family':'sans-serif', 
                            'xtick.labelsize':16, 
                            'ytick.labelsize':16, 
                            'figure.figsize':(16*(2/3), 9*(2/3)), 
                            'axes.labelsize':20
                           })
```

```python
dfa = pandas.read_csv('venn_a.tsv', sep='\t', index_col=0)
dfb = pandas.read_csv('venn_b.tsv', sep='\t', index_col=0)
dfc = pandas.read_csv('venn_c.tsv', sep='\t', index_col=0)
dfc
```

```python
seta = set(dfa.index)
setb = set(dfb.index)
setc = set(dfc.index)
```

```python
v = matplotlib_venn.venn3(
    subsets=[seta, setb, setc],
    set_labels = ('Set A', 'Set B', 'Set C'),
    set_colors=("skyblue", "gold", "tab:purple"),
)

for text in v.set_labels:
    text.set_fontsize(20)
for x in range(len(v.subset_labels)):
    if v.subset_labels[x] is not None:
        v.subset_labels[x].set_fontsize(20)
matplotlib.pyplot.show()
```
